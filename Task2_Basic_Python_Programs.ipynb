```python
def area_circle(radius):
    return 3.14 * radius * radius

print(area_circle(5))
```

```python
def largest(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= c:
        return b
    else:
        return c

print(largest(10, 25, 15))
```

```python
def string_length(s):
    count = 0
    for ch in s:
        count += 1
    return count

print(string_length("Python"))
```

```python
def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

print(is_prime(13))
```

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(6))
```

```python
def reverse_string(s):
    if s == "":
        return ""
    return s[-1] + reverse_string(s[:-1])

print(reverse_string("hello"))
```

```python
n = int(input("Enter n: "))
total = 0

for i in range(1, n + 1):
    total += i

print("Sum =", total)
```

```python
numbers = [1, 2, 3, 4, 5, 6, 8]

count = 0
for num in numbers:
    if num % 2 == 0:
        count += 1

print("Even numbers =", count)
```

```python
for num in range(2, 101):
    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        print(num, end=" ")
```
